The dataset `StudentsPerformance.csv` contains 1000 entries and 8 columns, with no missing values. Here's a precise description of its columns:

*   **gender**: (Object/Categorical) Indicates the gender of the student, e.g., 'female' or 'male'.
*   **race/ethnicity**: (Object/Categorical) Represents the student's ethnic group, such as 'group A', 'group B', etc.
*   **parental level of education**: (Object/Categorical) Describes the highest education level attained by the student's parents, e.g., 'bachelor's degree', 'some college', 'master's degree', 'associate's degree', 'high school', 'some high school'.
*   **lunch**: (Object/Categorical) Specifies the type of lunch provided to the student, e.g., 'standard' or 'free/reduced'.
*   **test preparation course**: (Object/Categorical) Indicates whether the student completed a test preparation course, e.g., 'none' or 'completed'.
*   **math score**: (Integer) The score obtained by the student in a math test.
*   **reading score**: (Integer) The score obtained by the student in a reading test.
*   **writing score**: (Integer) The score obtained by the student in a writing test.

```python
#Create colums transformer with 3 types of transformers

# Redefine features based on 'x' dataframe to ensure consistency
numerical_features=[feature for feature in x.columns if x[feature].dtype!='O']
categorical_features=[feature for feature in x.columns if x[feature].dtype=='O']

print(f"Numerical Features for ColumnTransformer: {numerical_features}")
print(f"Categorical Features for ColumnTransformer: {categorical_features}")

numerical_transformer=StandardScaler()
categorical_transformer=OneHotEncoder()

preprocessor=ColumnTransformer(
    [
        ('OneHotEncoder',categorical_transformer,categorical_features),
        ('StandardScaler',numerical_transformer,numerical_features)
    ]
)
```
